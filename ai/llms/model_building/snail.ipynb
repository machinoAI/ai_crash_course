```python
from datatrove.pipeline.readers import ParquetReader
from datatrove.utils._import_utils import check_required_dependencies
import inspect


check_required_dependencies("ParquetReader", ["pyarrow"])
```

```python
data_reader = ParquetReader("hf://datasets/HuggingFaceFW/fineweb-edu", glob_pattern="data/*/*.parquet")
```

```python
documents = data_reader.run()
```

```python
def load_dataset(documents, target_tokens = 1000):

    token_counts = 0

    for doc in documents:
        if token_counts <= target_tokens:
            
            if doc.metadata["language"]=="en":
                
                token_counts += doc.metadata["token_count"]
                text = doc.text
                print(text)

            if target_tokens >= token_counts:
                break
```

```python
load_dataset(documents)
```
