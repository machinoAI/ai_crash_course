| Pandas formula                         | Example                                                     | When to use                          |
| -------------------------------------- | ----------------------------------------------------------- | ------------------------------------ |
| `pd.DataFrame()`                       | `pd.DataFrame({"A":[1,2]})`                                 | Create a DataFrame                   |
| `df.head(n)`                           | `df.head(5)`                                                | View first rows                      |
| `df.tail(n)`                           | `df.tail(5)`                                                | View last rows                       |
| `df.shape`                             | `df.shape` → `(100, 5)`                                     | Get rows and columns                 |
| `df.columns`                           | `df.columns`                                                | Get column names                     |
| `df.dtypes`                            | `df.dtypes`                                                 | Check data types                     |
| `df.info()`                            | `df.info()`                                                 | Inspect columns, types, nulls        |
| `df.describe()`                        | `df.describe()`                                             | Summary statistics                   |
| `df["col"]`                            | `df["age"]`                                                 | Select one column                    |
| `df[["c1","c2"]]`                      | `df[["age","salary"]]`                                      | Select multiple columns              |
| `df.iloc[]`                            | `df.iloc[0, 2]`                                             | Select by integer position           |
| `df.loc[]`                             | `df.loc[5, "salary"]`                                       | Select by label                      |
| `df[df["age"] > 30]`                   | `df[df["age"] > 30]`                                        | Filter rows                          |
| `df[(c1) & (c2)]`                      | `df[(df.age>30) & (df.salary>50000)]`                       | Filter using AND                     |
| `df[(c1) \| (c2)]`                     | `df[(df.age<20) \| (df.age>60)]`                            | Filter using OR                      |
| `~condition`                           | `df[~(df["city"]=="Delhi")]`                                | Negate a condition                   |
| `df.query()`                           | `df.query("age > 30")`                                      | Cleaner filtering                    |
| `df["col"].unique()`                   | `df["city"].unique()`                                       | Find unique values                   |
| `df["col"].nunique()`                  | `df["user_id"].nunique()`                                   | Count unique values                  |
| `df["col"].value_counts()`             | `df["city"].value_counts()`                                 | Frequency of each value              |
| `df.isna()`                            | `df.isna()`                                                 | Find missing values                  |
| `df.isna().sum()`                      | `df.isna().sum()`                                           | Count missing values                 |
| `df.notna()`                           | `df["age"].notna()`                                         | Find non-missing values              |
| `df.dropna()`                          | `df.dropna()`                                               | Remove rows with missing values      |
| `df.dropna(subset=[])`                 | `df.dropna(subset=["age"])`                                 | Remove rows missing specific columns |
| `df.fillna()`                          | `df["age"].fillna(30)`                                      | Fill missing values                  |
| `df["col"].fillna(df["col"].mean())`   | `df["age"].fillna(df["age"].mean())`                        | Impute with mean                     |
| `df["col"].fillna(df["col"].median())` | `df["age"].fillna(df["age"].median())`                      | Impute with median                   |
| `df.drop_duplicates()`                 | `df.drop_duplicates()`                                      | Remove duplicate rows                |
| `df.drop_duplicates("col")`            | `df.drop_duplicates("user_id")`                             | Remove duplicates based on column    |
| `df.sort_values()`                     | `df.sort_values("salary")`                                  | Sort ascending                       |
| `df.sort_values(..., ascending=False)` | `df.sort_values("salary", ascending=False)`                 | Sort descending                      |
| `df.sort_values(["c1","c2"])`          | `df.sort_values(["dept","salary"])`                         | Sort by multiple columns             |
| `df["new"] = ...`                      | `df["annual"] = df["monthly"]*12`                           | Create a column                      |
| `df.rename()`                          | `df.rename(columns={"old":"new"})`                          | Rename columns                       |
| `df.drop(columns=[])`                  | `df.drop(columns=["age"])`                                  | Remove columns                       |
| `df["col"].apply()`                    | `df["age"].apply(lambda x:x+1)`                             | Apply a function                     |
| `df["col"].map()`                      | `df["gender"].map({"M":1,"F":0})`                           | Map individual values                |
| `df.replace()`                         | `df["gender"].replace({"M":1,"F":0})`                       | Replace values                       |
| `df.astype()`                          | `df["age"].astype(int)`                                     | Change data type                     |
| `df.groupby()`                         | `df.groupby("city")["salary"].mean()`                       | Aggregate by groups                  |
| `df.groupby().sum()`                   | `df.groupby("dept")["sales"].sum()`                         | Sum by group                         |
| `df.groupby().mean()`                  | `df.groupby("dept")["salary"].mean()`                       | Mean by group                        |
| `df.groupby().count()`                 | `df.groupby("dept")["id"].count()`                          | Count records per group              |
| `df.groupby().agg()`                   | `df.groupby("dept")["salary"].agg(["mean","max","min"])`    | Multiple aggregations                |
| `df.groupby().size()`                  | `df.groupby("dept").size()`                                 | Number of rows per group             |
| `df["col"].sum()`                      | `df["revenue"].sum()`                                       | Total                                |
| `df["col"].mean()`                     | `df["salary"].mean()`                                       | Average                              |
| `df["col"].median()`                   | `df["salary"].median()`                                     | Median                               |
| `df["col"].std()`                      | `df["salary"].std()`                                        | Standard deviation                   |
| `df["col"].min()`                      | `df["salary"].min()`                                        | Minimum                              |
| `df["col"].max()`                      | `df["salary"].max()`                                        | Maximum                              |
| `df["col"].quantile()`                 | `df["salary"].quantile(.95)`                                | Percentile                           |
| `df.corr()`                            | `df[["age","salary"]].corr()`                               | Correlation                          |
| `df["col"].rank()`                     | `df["salary"].rank(ascending=False)`                        | Rank values                          |
| `pd.merge()`                           | `pd.merge(df1,df2,on="user_id")`                            | Join two DataFrames                  |
| `pd.merge(..., how="left")`            | `pd.merge(df1,df2,on="id",how="left")`                      | Left join                            |
| `pd.merge(..., how="inner")`           | `pd.merge(df1,df2,on="id",how="inner")`                     | Inner join                           |
| `pd.concat()`                          | `pd.concat([df1,df2])`                                      | Combine DataFrames                   |
| `pd.concat(..., axis=1)`               | `pd.concat([df1,df2],axis=1)`                               | Combine columns                      |
| `df.pivot_table()`                     | `df.pivot_table(values="sales",index="city",aggfunc="sum")` | Reshape + aggregate                  |
| `df["col"].str.lower()`                | `df["name"].str.lower()`                                    | Convert strings to lowercase         |
| `df["col"].str.upper()`                | `df["name"].str.upper()`                                    | Convert strings to uppercase         |
| `df["col"].str.contains()`             | `df[df["name"].str.contains("raj")]`                        | Search within strings                |
| `df["col"].str.startswith()`           | `df[df["name"].str.startswith("A")]`                        | Filter by prefix                     |
| `df["col"].str.split()`                | `df["name"].str.split(" ")`                                 | Split strings                        |
| `pd.to_datetime()`                     | `df["date"]=pd.to_datetime(df["date"])`                     | Convert to datetime                  |
| `df["date"].dt.year`                   | `df["date"].dt.year`                                        | Extract year                         |
| `df["date"].dt.month`                  | `df["date"].dt.month`                                       | Extract month                        |
| `df["date"].dt.day`                    | `df["date"].dt.day`                                         | Extract day                          |
| `df["date"].dt.dayofweek`              | `df["date"].dt.dayofweek`                                   | Get weekday                          |
| `df.shift()`                           | `df["sales"].shift(1)`                                      | Previous/next row comparison         |
| `df.diff()`                            | `df["sales"].diff()`                                        | Difference between rows              |
| `df.cumsum()`                          | `df["sales"].cumsum()`                                      | Cumulative sum                       |
| `df.pct_change()`                      | `df["sales"].pct_change()`                                  | Percentage change                    |
| `df.reset_index()`                     | `df.reset_index(drop=True)`                                 | Reset index                          |
| `df.set_index()`                       | `df.set_index("user_id")`                                   | Set a column as index                |
| `df.sample()`                          | `df.sample(100)`                                            | Randomly sample rows                 |
| `df.nlargest()`                        | `df.nlargest(5,"salary")`                                   | Top N rows                           |
| `df.nsmallest()`                       | `df.nsmallest(5,"salary")`                                  | Bottom N rows                        |
